```python
# Step 1 — Split all skills into a list

all_skills =[]

for skills in df['tagsAndSkills'].dropna():
    for skill in skills.split(","):
        all_skills.append(skill.strip())
```

```python
print(all_skills)
```

```python
# Step 2 — Count frequency of each skill

from collections import Counter

skill_counts = Counter(all_skills)
```

```python
# Step 2 — Count frequency of each skill


print(skill_counts)
```

```python
# Step 3 — View top 20 most demanded skills

skill_counts.most_common(20)
```

# Skills Analysis Insights

1. **Python** is the most demanded skill with **9 job postings**.
2. **SQL** appears **7 times**, making database knowledge highly valuable.
3. **CSS** and **JavaScript** are among the top skills.
4. **Cloud** and **DevOps** skills like **AWS**, **Azure**, **CI/CD**, and **Continuous Integration** are frequently required.
5. Skills like **Java**, **Git**, **REST APIs**, and **MVC** also appear multiple times.

```python
# Step 4 — Plot Top 10 Skills 📊

top_skills = pd.DataFrame(
    skill_counts.most_common(10),
    columns=["Skill","Count"]
)

top_skills.plot(
    x = "Skill",
    y = "Count",
    kind = "barh",
    figsize=(10,6)
)

plt.xlabel=("Count")
plt.ylabel("Skills")
plt.ti("Top 10 Skills In Demanded Skills")
plt.gca().invert_yaxis()
plt.tight_layout()
plt,show()


```

# Experience Insight :

The experience analysis shows that most job postings require mid-level experience rather than entry-level experience.

The highest number of job postings are for 3–6 years and 3–5 years of experience, with 5 job postings each. This indicates that companies are actively hiring professionals with a few years of industry experience.

Roles requiring 5–10 years and 1–3 years also appear frequently, with 4 job postings each, showing demand for both experienced professionals and early-career candidates.

## Final Conclusion

This exploratory data analysis was performed on job postings related to Computer Engineering careers to understand hiring trends, required skills, salary patterns, and experience requirements in the current job market.

Key Findings
Software Development, Data Science, Cloud Engineering, DevOps, Testing, and Web Development were among the most common job roles identified in the dataset.
Major hiring activity was concentrated in specific locations, showing that technology opportunities are clustered around strong IT hubs.
Several companies appeared repeatedly in the dataset, indicating active hiring for computer engineering roles across multiple organizations.
Salary analysis showed variation across roles and locations, with higher salaries generally associated with specialized technical positions.
The skills analysis revealed that Python is the most frequently required skill, followed by SQL, JavaScript, CSS, Java, AWS, CI/CD, and Git. This highlights the importance of programming, database, web development, and cloud-related skills.
The word cloud visualization further confirmed that technical skills such as Python, SQL, cloud technologies, and development tools dominate job requirements.
Experience analysis showed that most openings require 3–6 years of experience, although opportunities for 1–3 years are also available, making the market accessible to early-career professionals as well.

## Overall Conclusion

Overall, the analysis shows that the job market for Computer Engineering students is highly focused on software development, programming, data handling, web technologies, and cloud computing. Students who build strong skills in Python, SQL, web development, and DevOps/cloud tools will have better employment opportunities and stronger career growth in the technology industry.

```python
import pandas as pd

query = """
SELECT
    title,
    ROUND((AVG(avg_salary)::numeric / 100000), 2) AS avg_salary_lpa
FROM jobs
WHERE avg_salary > 0
GROUP BY title
ORDER BY avg_salary_lpa DESC
LIMIT 10;
"""

salary_df = pd.read_sql(query, engine)
```

```python
import matplotlib.pyplot as plt

salary_df.plot(
    x='title',
    y='avg_salary_lpa',
    kind='barh',
    figsize=(12,6)
)

plt.xlabel("Job Role")
plt.ylabel("Average Salary (LPA)")
plt.title("Average Salary by Job Role")
# plt.gca().invert_yaxis()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
