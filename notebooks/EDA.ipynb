# Exploratory Data Analysis - Indian Tech Job Market Intelligence

This notebook analyzes hiring trends, required skills, salary distributions, and experience requirements in the 2025 Indian Tech Job Market.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from wordcloud import WordCloud

%matplotlib inline
sns.set_theme(style="darkgrid")
```

### Load the Cleaned Dataset

```python
df = pd.read_csv('../data/processed/jobs_cleaned.csv')
print(f"Dataset Shape: {df.shape}")
```

### Dataset Columns & Summary Stats

```python
df.info()
```

### 1. Job Roles Demand

```python
plt.figure(figsize=(12, 6))
df['title'].value_counts().head(10).plot(kind='barh', color='skyblue')
plt.title('Top 10 Most Demanded Job Roles')
plt.xlabel('Number of Openings')
plt.ylabel('Job Title')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### 2. Job Openings by Location

```python
plt.figure(figsize=(12, 6))
df['location'].value_counts().head(10).plot(kind='bar', color='salmon')
plt.title('Top 10 Cities for Computer Engineering Jobs')
plt.xlabel('Location')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 3. Salary Distribution Analysis

```python
df_sal = df[df['avg_salary'] > 0].copy()
df_sal['avg_salary_lpa'] = df_sal['avg_salary'] / 100000.0

plt.figure(figsize=(10, 5))
sns.histplot(df_sal['avg_salary_lpa'], kde=True, bins=25, color='teal')
plt.title('Salary Distribution in LPA (Lakhs Per Annum)')
plt.xlabel('Average Salary (LPA)')
plt.ylabel('Count')
plt.show()
```

### 4. Average Salary by Job Role

```python
plt.figure(figsize=(12, 6))
df_sal.groupby('title')['avg_salary_lpa'].mean().sort_values(ascending=False).head(10).plot(kind='barh', color='orchid')
plt.title('Top 10 Highest Paying Job Roles (LPA)')
plt.xlabel('Average Salary (LPA)')
plt.ylabel('Job Role')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### 5. Demanded Skills & Word Cloud

```python
all_skills = []
for skills in df['tagsAndSkills'].dropna():
    for skill in skills.split(','):
        all_skills.append(skill.strip())

skill_counts = Counter(all_skills)
print("Top 10 Most Demanded Skills:", skill_counts.most_common(10))

text = " ".join(all_skills)
wordcloud = WordCloud(width=1000, height=500, background_color='black', colormap='rainbow').generate(text)

plt.figure(figsize=(15, 8))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Most Demanded Skills Word Cloud')
plt.show()
```
